Build a decision tree classifier to predict whether a customer will purchase a product or service based on their demographic and behavioral data. Use a dataset such as the Bank Marketing dataset from the UCI Machine Learning Repository.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
bank_df = pd.read_csv('bank-additional-full.csv', sep=';')

# Display the first 5 rows
print('First 5 rows of the dataset:')
display(bank_df.head())
```

```python
# Display general information about the DataFrame
print('\nDataFrame Info:')
display(bank_df.info())
```

```python
# Display descriptive statistics for numerical columns
print('\nDescriptive statistics for numerical columns:')
display(bank_df.describe())
```

```python
# Identify categorical and numerical features
categorical_cols = bank_df.select_dtypes(include='object').columns.tolist()
numerical_cols = bank_df.select_dtypes(exclude='object').columns.tolist()

print(f"Categorical columns: {categorical_cols}")
print(f"Numerical columns: {numerical_cols}")
```

```python
# Preprocessing: Convert categorical variables to numerical using one-hot encoding
# Exclude the target variable 'y' for now
categorical_features_to_encode = [col for col in categorical_cols if col != 'y']

bank_df_encoded = pd.get_dummies(bank_df, columns=categorical_features_to_encode, drop_first=True)

# Convert the target variable 'y' to numerical (0 for 'no', 1 for 'yes')
bank_df_encoded['y'] = bank_df_encoded['y'].map({'no': 0, 'yes': 1})

print('\nFirst 5 rows of the encoded dataset:')
display(bank_df_encoded.head())

print('\nInfo of the encoded dataset:')
display(bank_df_encoded.info())
```

```python
# Split the data into features (X) and target (y)
X = bank_df_encoded.drop('y', axis=1)
y = bank_df_encoded['y']

print(f"Features (X) shape: {X.shape}")
print(f"Target (y) shape: {y.shape}")
print('\nFirst 5 rows of X:')
display(X.head())
print('\nFirst 5 values of y:')
display(y.head())
```

```python
# Split the dataset into training and testing sets
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")
```

```python
# Split the data into features (X) and target (y)
X = bank_df_encoded.drop('y', axis=1)
y = bank_df_encoded['y']

print(f"Features (X) shape: {X.shape}")
print(f"Target (y) shape: {y.shape}")
print('\nFirst 5 rows of X:')
display(X.head())
print('\nFirst 5 values of y:')
display(y.head())
```

```python
# Split the dataset into training and testing sets
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")
```

```python
# Import Decision Tree Classifier
from sklearn.tree import DecisionTreeClassifier

# Initialize the Decision Tree Classifier
# Using a random_state for reproducibility
dtc = DecisionTreeClassifier(random_state=42)

# Train the model
dtc.fit(X_train, y_train)

print("Decision Tree Classifier trained successfully!")
```

```python
# Make predictions on the test set
y_pred = dtc.predict(X_test)

print("Predictions made on the test set.")
```

```python
# Evaluate the model
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

print('\nAccuracy Score:')
print(accuracy_score(y_test, y_pred))

print('\nClassification Report:')
print(classification_report(y_test, y_pred))

print('\nConfusion Matrix:')
display(pd.DataFrame(confusion_matrix(y_test, y_pred),
                   index=['Actual No Purchase', 'Actual Purchase'],
                   columns=['Predicted No Purchase', 'Predicted Purchase']))
```
