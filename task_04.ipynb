## Task-04

Analyze and visualize sentiment patterns in social media data to understand public opinion and attitudes towards specific topics or brands.

Sample Dataset: https://github.com/Prodigy-InfoTech/data-science-datasets/tree/main/Task%204

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/Prodigy-InfoTech/data-science-datasets/refs/heads/main/Task%204/twitter_training.csv"

data = pd.read_csv(url, header=None)
# Rename the columns
data.rename(columns={2: 'sentiment', 3: 'text'}, inplace=True)

print('Sample Social Media Data with Sentiment:')
display(data.head())
```

```python
# Calculate sentiment counts explicitly
sentiment_counts = data['sentiment'].value_counts().reset_index()
sentiment_counts.columns = ['sentiment', 'count']

# Ensure the order for plotting is maintained
order = ['Positive', 'Neutral', 'Negative', 'Irrelevant']
sentiment_counts['sentiment'] = pd.Categorical(sentiment_counts['sentiment'], categories=order, ordered=True)
sentiment_counts = sentiment_counts.sort_values('sentiment')

# Visualize the distribution of sentiment using a barplot
plt.figure(figsize=(12, 7)) # Increased figure size for better label visibility
sns.barplot(x='sentiment', y='count', data=sentiment_counts, palette='viridis', hue='sentiment', legend=False)
plt.title('Sentiment Distribution in Social Media Data')
plt.xlabel('Sentiment')
plt.ylabel('Number of Posts')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout() # Use tight_layout first
plt.subplots_adjust(left=0.15) # Adjust left margin to prevent y-label truncation
plt.show()
```

```python
# Inspect unique values of the 'sentiment' column to identify actual labels
print("Unique sentiment values:")
print(data['sentiment'].unique())

# Also check the count of each unique value
print("\nSentiment value counts:")
print(data['sentiment'].value_counts())
```
